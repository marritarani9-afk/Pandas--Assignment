## Q1. Array operations and vectorization

**Predicted output**
```
[20 40 60]
0    20
1    40
2    60
dtype: int64
```

````python
import numpy as np
import pandas as pd

# Create a NumPy array
arr = np.array([10, 20, 30])

# Create a Pandas Series
s = pd.Series([10, 20, 30])

# Multiply each element of the NumPy array by 2
print(arr * 2)

# Multiply each element of the Pandas Series by 2
print(s * 2)
````

- **Difference:** A NumPy array is mainly used for numerical computation, while a Pandas Series has an index and is designed for labeled data.

## Q1. Broadcasting with shapes (3, 1) and (1, 3)

````python
import numpy as np

# Column array
a = np.array([[10], [20], [30]])

# Row array
b = np.array([[1, 2, 3]])

# Add arrays using broadcasting
result = a + b

print(result)

# Result shape
print(result.shape)

print("Shape of a:", a.shape)

print("Shape of b:", b.shape)
````

1. **Output:** a 3 × 3 grid `[[11, 12, 13], [21, 22, 23], [31, 32, 33]]`, shape `(3, 3)`.
2. **Shapes before addition:** `a` is `(3, 1)` and `b` is `(1, 3)`.
3. **How broadcasting works:** NumPy compares shapes from the last dimension backwards; two dimensions are compatible if they are equal or one of them is 1. A size-1 dimension is virtually stretched (no data is copied) to match the other: `a` is stretched across 3 columns and `b` across 3 rows, both become `(3, 3)`, and then they are added element-wise.
4. **Why it differs from adding two 1-D arrays:** two 1-D arrays of shape `(3,)` are added position by position (`[10,20,30] + [1,2,3] = [11,22,33]`, shape `(3,)`). Here every element of `a` is combined with every element of `b` (an "outer sum"), giving a 2-D grid.

**Part 4 — replacing `b` with `np.array([1, 2])`**

````python
import numpy as np

a = np.array([[10], [20], [30]])
b2 = np.array([1, 2])

print("Shape of b2:", b2.shape)
print(a + b2)
print((a + b2).shape)

# The case that actually fails: a 1-D array of length 3 plus a 1-D array of length 2
try:
    print(np.array([10, 20, 30]) + b2)
except ValueError as e:
    print("ValueError -", e)
````

**Explanation:** `b2` has shape `(2,)`, which broadcasts as `(1, 2)`; since `a` is `(3, 1)`, the dimensions 3 vs 1 and 1 vs 2 are both compatible, so it does **not** fail and returns a `(3, 2)` result `[[11, 12], [21, 22], [31, 32]]`. Broadcasting only raises an error when a pair of dimensions is unequal and neither is 1 (e.g. `(3,)` with `(2,)`, as shown above).

## Q2. Parsing a CSV with quoted commas, missing values and mixed date formats

````python
import pandas as pd

# Create a CSV file
with open("students.csv", "w") as f:
    f.write('ID,Name,Date,Marks\n'
            '101,"Asha, R",2026-09-01,80\n'
            '102,Ravi,01/09/2026,\n'
            '103,Neha,2026-09-03,NA\n')

# Read the CSV file
df = pd.read_csv("students.csv")

# Display the data
print(df)

# Display the data types
print(df.dtypes)
````

1. **Quoted comma:** `read_csv` treats text inside double quotes as a single field (`quotechar='"'`), so `"Asha, R"` stays one `Name` value with 4 columns intact. If the quotes were missing, the comma would split it into an extra column and shift the row.
2. **Date and Marks issues:**
   - `Date` has two formats (`2026-09-01` and `01/09/2026`), so it is loaded as plain text, not as dates, and sorting or date arithmetic would be wrong. `01/09/2026` is also ambiguous (1 September or 9 January?), so the intended format must be decided (assumed day/month/year below).
   - `Marks` contains a blank and `NA`; both silently become `NaN` by default, which forces the column to `float64` (80.0) and hides the fact that they may mean different things.

````python
import pandas as pd

# Keep "NA" as text (only truly blank cells become NaN) so it can be inspected first
df = pd.read_csv("students.csv", dtype=str, keep_default_na=False, na_values=[""])

# Inspect the "NA" entries before deciding what they mean
print(df[df["Marks"] == "NA"])

# Marks -> numeric; anything that is not a number ("NA", blank) becomes NaN
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# Dates -> datetime: parse each known format separately (assumed dd/mm/yyyy for the slash format)
iso = pd.to_datetime(df["Date"], format="%Y-%m-%d", errors="coerce")
dmy = pd.to_datetime(df["Date"], format="%d/%m/%Y", errors="coerce")
df["Date"] = iso.fillna(dmy)

print(df)
print(df.dtypes)
````

3. **Explanation:** `na_values=[""]` and `pd.to_numeric(..., errors="coerce")` turn missing or non-numeric marks into `NaN`, and parsing each date format explicitly (with `format=` and `fillna`) converts the mixed strings into one `datetime64` column without guessing.
4. **Why investigate `"NA"`:** it may mean "Not Available" (marks not entered), "Not Applicable" or "Absent" (the student did not take the exam), or even be a real value or a typing error; each needs different treatment (a student who was absent is not the same as missing data, and imputing or deleting blindly could bias the results), so its meaning should be confirmed with the data source first.

## Q3. Outliers: mean vs median

````python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())

# Part 4: record with the highest salary
print(df.nlargest(1, "Salary"))
````

1. **Prediction:** Mean = (25000 + 30000 + 35000 + 10000000) / 4 = **2522500.0**; Median = (30000 + 35000) / 2 = **32500.0**.
2. **Effect of the extreme value:** the single value 10,000,000 drags the mean up to about 78 times the median, higher than three of the four salaries by a huge margin, so the mean no longer describes anyone typical.
3. **Better statistic:** the **median (32,500)**, because it depends only on the middle values and is not affected by how large the outlier is; it sits close to the three ordinary salaries.
4. **Code:** `df.nlargest(1, "Salary")` (above) returns Employee D. **Explanation:** the record should be investigated before removal because it could be a data-entry error (e.g. extra zeros) or a genuine value (e.g. a CEO's salary); removing a genuine value would lose real information and distort the dataset, while keeping an error would distort the mean.

## Q1. Cleaning a dirty student assessment dataset

**Step 1 — Load and inspect the Dataset**

````python
import pandas as pd

# Load the tab-separated CSV file
df = pd.read_csv("student_assessment_dirty.csv", sep="\t")

# Display the first five rows
print(df.head())

print()

# Display dataset information
df.info()

print()

# Display missing values in each column
print(df.isnull().sum())
````

**Explanation:**

- `read_csv()` loads the student assessment CSV file into Pandas.
- `head()` and `info()` help us understand the data and its structure.
- `isnull()`.`sum()` shows the number of missing values in each column.

**Step 2 — Standardize department names**

````python
# Remove spaces and dots and convert department names to uppercase
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.replace(".", "", regex=False)
    .str.upper()
)

# Standardize the department names
df["Department"] = df["Department"].replace({
    "IT": "IT",
    "FINANCE": "Finance",
    "HR": "HR"
})

# Display unique department names
print(df["Department"].unique())
````

**Explanation:** 
- `strip()`, `replace()`, and `upper()` remove formatting differences.
Values such as `IT`, it, and `I.T`. are converted to `IT`.
Similarly, finance and Finance are standardized to Finance.

**Step 3 — Remove exact duplicate records** (done after standardizing, so `finance` vs `Finance` copies of the same record are also caught)

````python
# Find and display duplicate records
print("Duplicate records:")
print(df[df.duplicated()])

# Count the duplicate records
duplicates_removed = df.duplicated().sum()

# Remove exact duplicate records
df = df.drop_duplicates()

# Display the number of records removed
print("Records removed:", duplicates_removed)
````

**Explanation:** 
- `duplicated()` identifies repeated student - assessment records.
- `drop_duplicates()` removes the exact duplicate records.
- The duplicate count is stored so we can report it later.

**Steps 4  — Convert Marks to numeric format**

````python
# Convert Marks to string so text values can be cleaned
df["Marks"] = df["Marks"].astype(str)

# Remove the word "marks" from values such as "88 marks"
df["Marks"] = (
    df["Marks"]
    .str.replace("marks", "", case=False, regex=False)
    .str.strip()
)

# Convert the cleaned values to numbers
# Invalid values such as Absent and NA become NaN
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# Display the Marks column
print(df["Marks"])
````

**Explanation:**
- The text `"marks"` is removed from values such as `"88 marks"`.
- `pd.to_numeric()` converts valid mark values into numbers.
- Values such as `"Absent"` and `"NA"` become NaN instead of causing an error.

**Step 5 — Investigate Missing Marks**

````python
# Display records where marks are missing
print("Missing mark records:")
print(df[df["Marks"].isnull()][
    ["Student_ID", "Student_Name", "Marks"]
])

# Count missing marks
missing_marks = df["Marks"].isnull().sum()

print("Missing marks:", missing_marks)
````

**Explanation:** 
- `isnull()` identifies students whose marks are missing.
- We investigate these records before deciding whether to delete them.
- Deleting missing values without investigation may remove important information or introduce bias.

**Step 6 — Identify Invalid Marks**

````python
# Find marks outside the valid range of 0 to 100
invalid_marks = df[
    df["Marks"].notnull() &
    ~df["Marks"].between(0, 100)
]

# Display invalid marks
print("Invalid marks:")
print(invalid_marks[["Student_ID", "Student_Name", "Marks"]])

# Count invalid marks
print("Invalid marks identified:", len(invalid_marks))

# Replace invalid marks with missing values
df.loc[
    df["Marks"].notnull() &
    ~df["Marks"].between(0, 100),
    "Marks"
] = pd.NA
````

**Explanation:** 
- `between(0, 100)` checks whether marks are within the valid range.
The value `105` is identified as an invalid mark.
- Invalid marks are replaced with `NaN` because they need correction or verification.

**Step 7 - Display Cleaned Data and Summary**

````python
# Display the final cleaned DataFrame
print("Cleaned DataFrame:")
print(df)

print()

# Display the cleaning summary
print("Records removed:", duplicates_removed)
print("Missing marks remaining:", df["Marks"].isnull().sum())
print("Invalid marks identified:", len(invalid_marks))
````

**Explanation:** 
- The cleaned DataFrame is displayed after all cleaning operations.
- We report how many duplicate records were removed.
- We also report the remaining missing marks and invalid marks identified.
